# 1.3 Identifier & Grain Validation

## Row ID Uniqueness

`Row ID` is expected to uniquely identify each transaction-line record. We validate its uniqueness before using it as the record-level identifier.

````python
row_id_duplicates = sales_df["Row ID"].duplicated().sum()

print(f"Duplicate Row IDs: {row_id_duplicates:,}")
````

````python
print(f"Total records: {len(sales_df):,}")
print(f"Unique Row IDs: {sales_df['Row ID'].nunique():,}")
````

**Finding:** `Row ID` is unique across all 9,994 records, confirming that it can serve as the record-level identifier for the transactional dataset.

## Order-Level vs Transaction-Line Grain

`Order ID` is expected to repeat when a single order contains multiple products or transaction lines. We examine the distribution of transaction lines per order to validate the grain of the dataset.

````python
order_line_counts = sales_df.groupby("Order ID").size()

print(f"Unique Orders: {order_line_counts.size:,}")
print(f"Orders with multiple lines: {(order_line_counts > 1).sum():,}")
print(f"Maximum lines in a single order: {order_line_counts.max():,}")
````

````python
multi_line_orders = (
    order_line_counts[order_line_counts > 1]
    .sort_values(ascending=False)
)

display(multi_line_orders.head(10))
````

**Finding:** `Order ID` is not unique because individual orders can contain multiple transaction lines. Of the 5,009 orders, 2,471 contain more than one line, with a maximum of 14 lines per order. This confirms that the dataset is recorded at the transaction-line level rather than the order level.

## Customer Identifier Consistency

We validate whether each `Customer ID` consistently maps to a single `Customer Name`. This helps determine whether customer attributes can be reliably separated into a customer dimension during data modeling.

````python
customer_name_counts = (
    sales_df.groupby("Customer ID")["Customer Name"]
    .nunique()
)

inconsistent_customers = customer_name_counts[
    customer_name_counts > 1
]

print(f"Customers with multiple names: {len(inconsistent_customers):,}")
````

**Finding:** Each `Customer ID` maps consistently to a single `Customer Name`, with no conflicting customer-name mappings identified. This supports separating customer attributes into a dedicated customer dimension during data modeling.

## Product Identifier Consistency

We validate whether each `Product ID` consistently maps to a single `Product Name`. This check is important before creating a product dimension because inconsistent mappings could introduce duplicate or ambiguous product records.

````python
product_name_counts = (
    sales_df.groupby("Product ID")["Product Name"]
    .nunique()
)

inconsistent_products = product_name_counts[
    product_name_counts > 1
]

print(f"Products with multiple names: {len(inconsistent_products):,}")
````

### Investigate Inconsistent Product Mappings

The previous validation identified Product IDs associated with multiple Product Names. We inspect the affected records to determine whether the inconsistencies represent data-quality issues or legitimate product variations.

````python
# Identify Product IDs associated with multiple Product Names
inconsistent_product_ids = inconsistent_products.index

product_mapping_issues = (
    sales_df[
        sales_df["Product ID"].isin(inconsistent_product_ids)
    ][
        ["Product ID", "Product Name", "Category", "Sub-Category"]
    ]
    .drop_duplicates()
    .sort_values(["Product ID", "Product Name"])
)

display(product_mapping_issues.head(20))
````

### Investigate the pattern

````python
example_product_id = inconsistent_product_ids[0]

display(
    sales_df[
        sales_df["Product ID"] == example_product_id
    ][
        [
            "Order ID",
            "Order Date",
            "Product ID",
            "Product Name",
            "Category",
            "Sub-Category",
            "Sales",
            "Quantity",
            "Profit"
        ]
    ].sort_values("Order Date")
)
````

**Finding:** 32 Product IDs were identified with multiple Product Name mappings. A detailed inspection of `FUR-BO-10002213` showed that two different product names occur under the same Product ID across overlapping periods, indicating a source-data integrity issue rather than a simple historical product-name change. The raw values will be preserved, and the appropriate treatment will be determined during the data preparation and modeling stages.

### Business Impact of Product Mapping Issues

We quantify the sales and profit associated with Product IDs that have inconsistent Product Name mappings to assess the potential business impact of the source-data issue.

````python
# Total sales and profit
total_sales = sales_df["Sales"].sum()
total_profit = sales_df["Profit"].sum()

# Records associated with inconsistent Product IDs
affected_sales_df = sales_df[
    sales_df["Product ID"].isin(inconsistent_product_ids)
]

affected_sales = affected_sales_df["Sales"].sum()
affected_profit = affected_sales_df["Profit"].sum()

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Affected Sales: ${affected_sales:,.2f}")
print(f"Affected Sales %: {(affected_sales / total_sales) * 100:.2f}%")

print(f"\nTotal Profit: ${total_profit:,.2f}")
print(f"Affected Profit: ${affected_profit:,.2f}")
print(f"Affected Profit %: {(affected_profit / total_profit) * 100:.2f}%")
````

**Business Impact:** The 32 Product IDs with inconsistent Product Name mappings are associated with $97,465.67 in sales, representing 4.24% of total sales. These records also account for $13,230.92 in profit, or 4.62% of total profit. The issue is therefore material enough to document and consider during data modeling, but it does not invalidate the dataset for overall commercial analysis.

## Customer Attribute Consistency

We validate whether customer-level attributes remain consistent for each Customer ID. This helps identify potential dimensional integrity issues before creating the customer dimension.

````python
customer_attributes = [
    "Customer Name",
    "Segment",
    "Country",
    "City",
    "State",
    "Postal Code",
    "Region"
]

customer_consistency = (
    sales_df.groupby("Customer ID")[customer_attributes]
    .nunique()
)

inconsistent_customer_attributes = (
    customer_consistency > 1
).sum()

display(
    inconsistent_customer_attributes[
        inconsistent_customer_attributes > 0
    ]
    .sort_values(ascending=False)
    .rename("Customers with Multiple Values")
    .to_frame()
)
````

### Investigating Customer Location Variations

Several customer-level geographic attributes vary across transactions for the same Customer ID. We investigate individual customers to determine whether these fields represent customer master attributes or transaction-level locations.

````python
# Find a customer with multiple cities
customers_with_multiple_cities = (
    customer_consistency[
        customer_consistency["City"] > 1
    ].index
)

example_customer_id = customers_with_multiple_cities[0]

display(
    sales_df[
        sales_df["Customer ID"] == example_customer_id
    ][
        [
            "Order ID",
            "Order Date",
            "Customer ID",
            "Customer Name",
            "City",
            "State",
            "Postal Code",
            "Region"
        ]
    ].sort_values("Order Date")
)
````

**Finding:** Customer geographic attributes are not stable across transactions. For example, Customer ID `AA-10315` appears across multiple cities, states, postal codes, and regions during the analysis period. This indicates that these geographic fields represent transaction-level/order-location information rather than fixed customer master attributes. Therefore, they should not be treated as permanent attributes of the customer dimension during data modeling.

## Shipping Duration Validation

We calculate the number of days between Order Date and Ship Date to identify potentially abnormal shipping durations.

````python
shipping_days = (
    sales_df["Ship Date"] - sales_df["Order Date"]
).dt.days

shipping_summary = shipping_days.agg(
    ["min", "max", "mean", "median"]
)

display(shipping_summary)
````

````python
long_shipments = sales_df.loc[
    shipping_days > 30,
    [
        "Order ID",
        "Order Date",
        "Ship Date",
        "Ship Mode",
        "Sales",
        "Profit"
    ]
].copy()

long_shipments["Shipping Days"] = shipping_days[
    shipping_days > 30
]

print(f"Shipments taking more than 30 days: {len(long_shipments):,}")

display(
    long_shipments.sort_values(
        "Shipping Days",
        ascending=False
    ).head(10)
)
````

````python
shipping_distribution = (
    shipping_days
    .value_counts()
    .sort_index()
)

display(shipping_distribution.head(20))
````

````python
print("Shipping days frequency:")
display(
    shipping_distribution
    .sort_values(ascending=False)
    .head(30)
)
````

````python
shipping_by_mode = (
    sales_df.assign(
        Shipping_Days=shipping_days
    )
    .groupby("Ship Mode")["Shipping_Days"]
    .agg(
        ["count", "min", "median", "mean", "max"]
    )
)

display(shipping_by_mode)
````

## Establish the Normal Shipping Pattern

Let's calculate the distribution for the short-duration transactions first.

````python
shipping_mode_normal = (
    sales_df.assign(
        Shipping_Days=shipping_days
    )
    .loc[lambda df: df["Shipping_Days"] <= 7]
    .groupby("Ship Mode")["Shipping_Days"]
    .agg(
        ["count", "min", "median", "mean", "max"]
    )
)

display(shipping_mode_normal)
````

````python
shipping_mode_compliance = (
    sales_df.assign(
        Shipping_Days=shipping_days
    )
    .groupby("Ship Mode")
    .agg(
        Total_Records=("Shipping_Days", "size"),
        Within_7_Days=("Shipping_Days", lambda x: (x <= 7).sum())
    )
)

shipping_mode_compliance["Within_7_Days_%"] = (
    shipping_mode_compliance["Within_7_Days"]
    / shipping_mode_compliance["Total_Records"]
    * 100
)

display(shipping_mode_compliance)
````

**Finding:** Shipping durations show a strong concentration within 0–7 days across all shipping modes, with medians ranging from 0 to 4 days. However, a substantial number of records contain much longer durations, including repeated clusters around approximately 30, 60, 90, 120, 150, 180, and 210+ days. The maximum duration is 214 days. These patterns are inconsistent with the dominant operational shipping pattern and indicate a potential systematic issue in the source Ship Date values. The dates will be investigated further before being used for shipping-performance analysis.

## Anomaly in Shipping Dates

Let's test whether anomalous Ship Dates preserve the same day of month as Order Dates.

````python
anomalous_shipping = sales_df.assign(
    Shipping_Days=shipping_days
)

anomalous_shipping = anomalous_shipping[
    anomalous_shipping["Shipping_Days"] > 7
].copy()

anomalous_shipping["Same_Day_of_Month"] = (
    anomalous_shipping["Order Date"].dt.day
    ==
    anomalous_shipping["Ship Date"].dt.day
)

print(
    f"Anomalous shipments: {len(anomalous_shipping):,}"
)

print(
    f"Same day-of-month: "
    f"{anomalous_shipping['Same_Day_of_Month'].sum():,}"
)

print(
    f"Same day-of-month %: "
    f"{anomalous_shipping['Same_Day_of_Month'].mean() * 100:.2f}%"
)
````

**Final Assessment:** 3,013 transactions have shipping durations greater than 7 days, and 83.31% of these anomalous records preserve the same day of month between Order Date and Ship Date. Combined with the repeated clusters of approximately 30, 60, 90, 120, 150, 180, and 210+ days, this indicates a systematic anomaly in the source Ship Date values rather than a reliable representation of actual delivery duration.

The raw Ship Date will therefore be preserved without arbitrary correction. A derived shipping-duration field may be created later with an explicit anomaly flag, but delivery-performance KPIs will not be based on unvalidated corrected dates.

# 2. Transformation Strategy

The raw datasets will be transformed into a star-schema analytical model.

# 2.1 Target Model

- **FactSales:** Transaction-line-level sales records and numeric business measures.
- **DimDate:** One record per calendar date for time-based analysis.
- **DimCustomer:** Stable customer attributes including Customer ID, Customer Name, and Segment.
- **DimProduct:** Product identifier and descriptive product attributes, with known source-data mapping inconsistencies documented.
- **DimGeography:** Transaction-level geographic attributes including City, State, Postal Code, and Region.

### Transformation Responsibilities

**Python**
- Source-data profiling
- Data-quality investigation
- Structural validation

**Power Query**
- Data type enforcement
- Column selection and renaming
- Derived fields
- Dimension construction
- Data preparation for the analytical model

**DAX**
- Business measures
- Time intelligence
- Ratios and profitability metrics
- Growth and ranking calculations

The objective is to keep transformations in the layer where they are most appropriate rather than performing all preparation in a single tool.
